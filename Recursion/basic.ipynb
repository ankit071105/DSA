## Advance Recursion & Backtracking 

- Print  all the subsequences
- Find all subsequences with sum = K
- Check if there exists a subsequences with sum ==k
- Count all the subsequences with  sum ==k

```python
# Generate All Subsequences Using Recursion
def subsets(nums):
        r = []
        def fun(i, s): # index and a subset list 
            if i >= len(nums): 
                r.append(s[:]) # copy
                return 
            s.append(nums[i])
            fun(i + 1, s) # recursively call
            s.pop()
            fun(i + 1, s)  # recursively call
        fun(0, [])
        return r
nums = [1,2,3]
print(subsets(nums))
        
```

```python
# Generate Subsequences with Sum K Using Recursion
def Subsequenceswithk(nums, target):
    r = []

    def solve(i, t, s):
        if t == target:
            r.append(s[:])
            return
        if t > target:
            return
        if i >= len(nums):
            return
        s.append(nums[i])
        sums = t + nums[i]
        solve(i + 1, sums, s)
        s.pop()
        solve(i + 1, t, s)
        
    solve(0, 0, [])
    return r

nums = [1, 2, 3]
target = 5
print(Subsequenceswithk(nums, target))
    
        
            
            
```

```python
# Check if a Subsequence with Sum = K Exists 
def checkSubsequenceSum(arr, k):
        def solve(i,t,s):
            # base condtions 
            if t==k:
                return True
            elif t>k:
                return False
            if i >= len(arr):
                return False
            
            s.append(arr[i])
            Sums = t + arr[i]
            
            if solve(i+1,Sums,s):
                return True
            s.pop()
    
            if solve(i+1 , t,s):
                return True
            return False
        return solve(0, 0, [])
    
arr = [10, 1, 2, 7, 6, 1, 5]
k = 8
print(checkSubsequenceSum(arr,k))
```

```python
# Count all the subsequences with  sum ==k
def perfectSum(arr, target):
        def solve(i, t):
            if i >= len(arr):
                if t ==target:
                    return 1
                return 0
            p = solve(i + 1, t + arr[i])
            n = solve(i + 1, t)
            return p + n
        return solve(0, 0)
    
arr = [5, 2, 3, 10, 6, 8]
target = 10
print(perfectSum(arr,target))
```

```python
# Generate all Binary String
def binstr(n):
        num=[""]*n
        r=[]
        def solve(i,n,res):
            # base condition
            if i >= len(n):
                res.append("".join(n))
                return 

        # call recursively
            n[i]="0"
            solve(i+1,n,res)
           
            n[i]="1"
            solve(i+1,n,res)
    
    # call function
        solve(0,num,r)
        return r
print(binstr(2))
```

```python
#  Generate Parentheses
def generateParenthesis(n):
        b=[""]*(n*2)
        r=[]
        def solve(i,t):
            if i >= len(b):
                if t == 0:
                    r.append("".join(b))
                return 
            if t > len(b)//2:
                return
            elif t <0:
                return
            b[i]="("
            solve(i+1,t+1)
            b[i]=")"
            solve(i+1,t-1)
        solve(0,0)
        return r
print(generateParenthesis(2))
```

```python
# Combination Sum
def combinationSum(candidates, target):
        r=[]
        def solve(i,t,s,n):
            if t==target:
                r.append(s[:])
                return
            elif t >target:
                return
            if i >= len(n):
                return
            
            d=t+n[i]
            s.append(n[i])
            solve(i,d,s,n)
            d=t
            s.pop()
            solve(i+1,d,s,n)
        solve(0,0,[],candidates)
        return r
c = [2,3,6,7]
t = 7
print(combinationSum(c,t))
```

```python
# Combination Sum II 
def combinationSum2(c, target):
        r = []
        c.sort()
        def solve(i, t, s):
            if t == 0:
                r.append(s[:])
                return
            if t < 0:
                return
            for j in range(i, len(c)):
                if j > i and c[j] == c[j - 1]:
                    continue
                if c[j] > t:
                    break
                s.append(c[j])
                solve(j + 1, t - c[j], s)
                s.pop()
                
        solve(0, target, [])
        return r
c = [10,1,2,7,6,1,5]
t = 8
print(combinationSum2(c,t))
```

```python
# Combination Sum III
def combinationSum3(k, n):
        r=[]
        def solve(l,t,s):
            if t==n and len(s) ==k:
                r.append(s[:])
                return 
            
            if t > n or len(s) >k:
                return
            
            for i in range(l,10):
                d=t+i
                s.append(i)
                solve(i+1,d,s)
                s.pop()
        solve(1,0,[])
        return r
k = 3
n = 7
print(combinationSum3(k,n))
```

```python
# Letter Combinations of a Phone Number 
def letterCombinations(self, digits):
        c={"2":"abc", "3":"def" , "4":"ghi" , "5":"jkl" , "6":"mno", "7":"pqrs","8":"tuv","9":"wxyz"}
        r=[]
        def solve(i,s):
            if i >=len(digits):
                r.append("".join(s))
                return
            for j in c[digits[i]]:
                s.append(j)
                solve(i+1,s)
                s.pop()
        solve(0,[])
        return r
```

```python
# Python - N-Queens Problem 
def solveNQueens(n):
        board=['.'*n for _ in range(n)]
        ans=[]
        def issafe(r,c):

            # Upper-left diagonal
            x=r
            y=c
            while r >= 0 and c>=0:
                if board[r][c] == 'Q':
                    return False
                r-=1
                c-=1

            # Same row, left side
            r=x
            c=y
            while c>=0:
                if board[r][c] == 'Q':
                    return False
                c -=1

            # Lower-left diagonal
            r=x
            c=y
            while r< n and c >=0:
                if board[r][c] == 'Q':
                    return False
                r+=1
                c-=1
            return True
        
        def solve(c):
            if c==n:
                ans.append(list(board))
                return
            
            for r in range(n):
                if issafe(r,c):
                    board[r]=board[r][:c] + 'Q'+ board[r][c+1 :] #Place the Queen
                    solve(c+1) # Go to the next column
                    board[r]=board[r][:c]+ '.'+ board[r][c+1 :] #Remove the Queen(backtracking)
        solve(0)
        return ans
print(solveNQueens(4))
```

```python
# Rat in a Maze Problem
def ratInMaze(self, maze: list[list[int]]) -> list[str]:
        r=len(maze)
        c=len(maze[0])  
        a=[]
        v = [[0] * c for _ in range(r)]
        
        def find(i,j,m):
            # destination
            if i == r-1 and j== c-1:
                a.append(m)
                return
            v[i][j]=1
            #  (DLRU - LEXICAL ORDER)
            # down
            if i + 1 < r and not v[i+1][j] and maze[i+1][j]==1:
                find(i+1, j, m+"D")
            # left
            if j-1 >= 0 and not v[i][j-1] and maze[i][j-1]==1:
                find(i, j-1, m+"L")
                
            # Right
            if j+1 < c and not v[i][j+1] and maze[i][j+1]==1:
                find(i, j+1 , m+"R")
            # upward 
            if i - 1 >= 0 and not v[i-1][j] and maze[i-1][j]==1:
                find(i-1 , j , m+"U")
            
            
            v[i][j]=0
                
        if maze[0][0] ==1:
            find(0,0,"")
        
        return a
```
