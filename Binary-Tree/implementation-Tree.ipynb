```python
"""
        drinks
        /    \
     hot      cold
    /   \     /   \
  tea  coffee cola fanta
"""

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

drinks = Node("drinks")
hot= Node("hot")
cold= Node("cold")
tea = Node("tea")
coffee = Node("coffee")
cola = Node("cola")
fanta = Node("fanta")


# first Node
drinks.left = hot
drinks.right = cold

# second Node
hot.left = tea
hot.right = coffee

# third Node
cold.left = cola
cold.right = fanta



# left and right data of drinks
print(drinks.left.data)
print(drinks.right.data)


# left and right data of hot
print(hot.left.data)
print(hot.right.data)

# left and right data of cold
print(cold.left.data)
print(cold.right.data)

#full top to bottom left
print(drinks.left.left.data)

#full top to bottom right
print(drinks.right.right.data)
```
