```python
#  Binary Tree BFS traversal
from collections import deque
def levelOrder(self, root):
        if not root:
            return []
        r = []
        q = deque([root])
        while len(q) !=0:     
                e = q.popleft()
                r.append(e.val)
                if e.left:
                    q.append(e.left)
                if e.right:
                    q.append(e.right)
        return r
```

```python
# 102. Binary Tree Level Order Traversal 
from collections import deque
class Solution(object):
    def levelOrder(self, root):
        if not root:
            return []
        r = []
        q = deque([root])
        while q:
            level = []         
            n = len(q)          
            for _ in range(n):
                e = q.popleft()
                level.append(e.val)
                if e.left:
                    q.append(e.left)
                if e.right:
                    q.append(e.right)
            r.append(level)
        return r
```
