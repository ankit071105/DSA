```python
#  Binary Tree BFS traversal
from collections import deque
def levelOrder(self, root):
        if not root:
            return []
        r = []
        q = deque([root])
        while len(q) !=0:     
                e = q.popleft()
                r.append(e.val)
                if e.left:
                    q.append(e.left)
                if e.right:
                    q.append(e.right)
        return r
```

```python
# 102. Binary Tree Level Order Traversal 
from collections import deque
class Solution(object):
    def levelOrder(self, root):
        if not root:
            return []
        r = []
        q = deque([root])
        while q:
            level = []         
            n = len(q)          
            for _ in range(n):
                e = q.popleft()
                level.append(e.val)
                if e.left:
                    q.append(e.left)
                if e.right:
                    q.append(e.right)
            r.append(level)
        return r
```

```python
 # Top View of Binary Tree
 
#  Input: root = [1, 2, 3]
#  Output: [2, 1, 3]

from collections import deque
class Solution:
    def topView(self, root):
        if not root:
            return None
        a=[]
        q=deque()
        r={}
        q.append((root,0))
        while q:
            e,l=q.popleft()
            if l not in r:
                r[l] = e.data
            if e.left:
                q.append((e.left, l-1))
            if e.right:
                q.append((e.right,l+1))
        for v in sorted(r.items()):
            a.append(v[1])
        return a
```

```python
# Bottom View of Binary Tree 
# difference between top view and bottom view is :
# if condition removes in bottomm view code we direclty append the value of node to the dictionary with its level as key.

# Input: root = [1, 2, 3, 4, 5, N, 6]
# Output: [4, 2, 5, 3, 6]


from collections import deque
class Solution:
    def bottomView(self, root):
        # code here
        if not root:
            return None
        a=[]
        q=deque()
        r={}
        q.append((root , 0))
        while q:
            e,l = q.popleft()
            r[l] = e.data
            if e.left:
                q.append((e.left,l-1))
            if e.right:
                q.append((e.right, l+1))
                
        for v in sorted(r.items()):
            a.append(v[1])
        return a
```

```python
# 199. Binary Tree Right Side View

# Input: root = [1,2,3,null,5,null,4]
# Output: [1,3,4]


def rightSideView(self, root):
        if not root:
            return []

        q=deque()
        r=[]
        q.append(root)
        while len(q) != 0:
            l=len(q)
            for i in range(l):
                n=q.popleft()
                if i == l-1: # change in this conditon for right view
                    r.append(n.val)
                if n.left:
                    q.append(n.left)
                if n.right:
                    q.append(n.right)
        return r
```

```python
# Left View of Binary Tree


# Input: root = [1, 2, 3, 4, 5, N, N] 
# Output:[1, 2, 4]

from collections import deque
class Solution:
    def leftView(self, root):
        if not root:
            return []
        
        q=deque()
        r=[]
        q.append(root)
        while len(q) !=0:
            l=len(q)
            for i in range(l):
                n=q.popleft()
                if i==0: # change in this conditon for left view
                    r.append(n.val)
                if n.left:
                    q.append(n.left)
                if n.right:
                    q.append(n.right)
        return r
```
