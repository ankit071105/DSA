```python
# 144. Binary Tree Preorder Traversal


def preorderTraversal(self, root):
        a=[]
        def preorder(node):
            if not node:
                return
            a.append(node.val)
            preorder(node.left)
            preorder(node.right)
        preorder(root)
        return a
        
```

```python
# 94. Binary Tree Inorder Traversal

def inorderTraversal(self, root):
        a=[]
        def inorder(node):
            if not node:
                return 
            inorder(node.left)
            a.append(node.val)
            inorder(node.right)
        inorder(root)
        return a
```

```python
# 100. Same Tree

# Input: p = [1,2,3], q = [1,2,3]
# Output: true

def isSameTree(self, p, q):
        if not p and not q:
            return True
        
        if not p or not q:
            return False

        return (p.val == q.val and self.isSameTree(p.left, q.left) and  self.isSameTree(p.right, q.right))
```

```python
# 101. Symmetric Tree

# Input: root = [1,2,2,3,4,4,3]
# Output: true

def isSymmetric(self, root):
        def check(l, r):
            if not l or not r:
                return l == r
            return (l.val == r.val and check(l.left, r.right) and check(l.right, r.left))

        return check(root.left, root.right)
```

```python
# 104. Maximum Depth of Binary Tree

# Input: root = [3,9,20,null,null,15,7]
# Output: 3

def maxDepth(self, root):
        if not root:
            return 0
        l=self.maxDepth(root.left)
        r=self.maxDepth(root.right)
        return 1+max(l,r)
```

```python
# 111. Minimum Depth of Binary Tree

# Input: root = [3,9,20,null,null,15,7]
# Output: 2

def minDepth(self, root):
        if not root:
            return 0
        l=self.minDepth(root.left)
        r=self.minDepth(root.right)

        if l == 0:
            return 1 + r
        if r == 0:
            return 1 + l
            
        return 1+min(l,r)
```

```python
# 543. Diameter of Binary Tree

# Input: root = [1,2,3,4,5]
# Output: 3

def diameterOfBinaryTree(self, root):
        self.d=0
        def height(root):
            if not root:
                return 0
            l=height(root.left)
            r=height(root.right)
            self.d=max(self.d,l+r)
            return 1+max(l,r)
        height(root)
        return self.d
```

```python
# 572. Subtree of Another Tree
# Input: root = [3,4,5,1,2], subRoot = [4,1,2]
# Output: true
# 572. Subtree of Another Tree
# Input: root = [3,4,5,1,2], subRoot = [4,1,2]
# Output: true
        
        
def isSubtree(self, root, subRoot):
        def same(p, q):
            if not p and not q:
                return True
            if not p or not q:
                return False
            return (p.val == q.val and same(p.left, q.left) and same(p.right, q.right))
       
        if not root:
            return False
        if same(root, subRoot):
            return True
        return (self.isSubtree(root.left, subRoot) or self.isSubtree(root.right, subRoot))
```

```python
# 110. Balanced Binary Tree 

# Input: root = [3,9,20,null,null,15,7]
# Output: true

class Solution(object):
    def isBalanced(self, root):
        
        def solve(root):
            if root is None:
                return 0
            l=solve(root.left)
            if l==-1:
                return -1
            r=solve(root.right)
            if r==-1:
                return -1
            if abs(l-r) >1:
                return -1
            return 1+max(l,r)
        s=solve(root)
        return s != -1
```

```python
# 113. Path Sum II

# Input: root = [10,5,-3,3,2,null,11,3,-2,null,1], targetSum = 8
# Output: 3


class Solution(object):
    def pathSum(self, root, targetSum):
        s = []
        def dfs(r, t, p):
            if not r:
                return
            p.append(r.val)
            if not r.left and not r.right:
                if r.val == t:
                    s.append(p[:])
            else:
                a = t - r.val
                dfs(r.left, a, p)
                dfs(r.right, a, p)
            p.pop()
        dfs(root, targetSum, [])
        return s
    
    
    
```

```python

# 222. Count Complete Tree Nodes

# Input: root = [1,2,3,4,5,6]
# Output: 6

class Solution(object):
    def countNodes(self, root):
        if not root:
            return 0
        
        l = self.getDepth(root.left)
        r = self.getDepth(root.right)
        
        if l == r:
            return (1 << l) + self.countNodes(root.right)
        else:
            return (1 << r) + self.countNodes(root.left)
    
    def getDepth(self, node):
        d = 0
        while node:
            d += 1
            node = node.left
        return d
```
