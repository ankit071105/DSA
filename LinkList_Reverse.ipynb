```python
# 206. Reverse Linked List

class Solution(object):
    def reverseList(self, head):
        p=None
        c=head
        while c is not None:
            n=c.next
            c.next=p
            p=c
            c=n
        return p
```

```python
# 206. Reverse Linked List


class Solution(object):
    def reverseList(self, head):
        
    # convert linked list to array list  
        a=[]
        c=head
        while c:
            a.append(c.val)
            c=c.next
    # reverse logic
        a=a[::-1]
        
    # converting to array list to Linked list 
        i=0
        c=head
        while c:
            c.val=a[i]
            c=c.next
            i+=1
        return head
```

```python
class Solution(object):
    def isPalindrome(self, head):
        
    # convert linked list to array list  
        a = []
        while head is not None:
            a.append(head.val)
            head = head.next
        
    # palindrome logic 
        return a == a[::-1]
```

```python
# 2130. Maximum Twin Sum of a Linked List

class Solution(object):
    def pairSum(self, head):

      # convert linked list to array list 
        a=[]
        while head is not None:
            a.append(head.val)
            head=head.next
        
    # applying pair sum logic
        s=m=0
        n=len(a)
        for i in range(n//2):
            s=a[i]+a[n-1-i]
            m=max(m,s)
        return m
```

```python
# 2130. Maximum Twin Sum of a Linked List two pointer approach

class Solution(object):
    def pairSum(self, head):
        
        
        # convert linked list to array list 
        a=[]
        while head is not None:
            a.append(head.val)
            head=head.next
            
        # applying pair sum logic
        l=s=0
        r=len(a)-1
        while l<r:
            s= max(s,a[r]+a[l])
            l+=1
            r-=1
        return s
```

```python
# 92. Reverse Linked List II  (m to n)
class Solution(object):
    def reverseBetween(self, head, left, right):
    
     # converting to linked list to array list 
        a = []
        c = head
        while c:
            a.append(c.val)
            c = c.next
            
            
        # Reverse List  between m to n 
        left -= 1
        right -= 1
        while left < right:
            a[left], a[right] = a[right], a[left]
            left += 1
            right -= 1

        # converting to array list to Linked list 
        c = head
        i = 0
        while c:
            c.val = a[i]
            c = c.next
            i += 1
        return head
```

```python
# 24. Swap Nodes in Pairs
class Solution(object):
    def swapPairs(self, head):
    
    # converting to linked list to array list 
        a=[]
        c = head
        while c:
            a.append(c.val)
            c=c.next
            
     # applying swap logic for pairs 
        i=0
        while i < len(a) - 1:
            a[i],a[i+1]=a[i+1],a[i]
            i+=2
            
    # converting to  array list  to linked list 
        c = head
        i = 0
        while c:
            c.val = a[i]
            c = c.next
            i += 1
        return head
```

```python
# Rotate List (circular linkedlist)
class Solution(object):
    def rotateRight(self, head, k):

      # converting to linked list to array list 
        a=[]
        c=head
        while c:
            a.append(c.val)
            c=c.next
          
        # performing rotation of list   
        n=len(a)
        b=[0]*n
        for i in range(n):
            b[(i+k)%n]=a[i]
            
        # converting back to linklist 
        c=head
        i=0
        while c:
            c.val=b[i]
            c=c.next
            i+=1
        return head
        
```

```python
# 25. Reverse Nodes in k-Group

class Solution(object):
    def reverseKGroup(self, head, k):
        
        # converting to linked list to array list 
        a=[]
        c=head
        while c:
            a.append(c.val)
            c=c.next
        
        # Reverse logic for k-Group
        n=len(a)
        for i in range(0,n,k):
            if i + k <= n:
                l = i
                r = i + k - 1
            while l<r:
                a[l],a[r] =a[r],a[l]
                l+=1
                r-=1
        
        # array list to linkedlist
        c=head
        i=0
        while c:
            c.val=a[i]
            c=c.next
            i+=1
        return head
```
