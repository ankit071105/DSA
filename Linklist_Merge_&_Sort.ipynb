```python
# Merge Two Sorted Lists
class Solution(object):
    def mergeTwoLists(self, list1, list2):
        
        # linkedlist to array
        a = []; b = []
        c = list1
        while c:
            a.append(c.val)
            c = c.next
        c = list2
        while c:
            b.append(c.val)
            c = c.next

        # merge 
        arr = a + b
        # sort 
        arr.sort()

        #array to linkedlist
        d = ListNode(0)
        c = d
        for i in range(len(arr)):
            c.next = ListNode(arr[i])
            c = c.next
        return d.next
```

```python

# Remove Duplicates from Sorted List
class Solution(object):
    def deleteDuplicates(self, head):
        c = head
        while c and c.next:
            if c.val == c.next.val:
                c.next = c.next.next
            else:
                c = c.next
        return head
```

```python

# Sort List
class Solution(object):
    def sortList(self, head):
    
    # convert linked list to list 
        a=[]
        c=head
        while c:
            a.append(c.val)
            c=c.next
            
    # applying sorting
        a.sort()
        
    # convert list to linked list 
        c=head
        i=0
        while c:
            c.val=a[i]
            c=c.next
            i+=1
        return head
```

```python
# 143. Reorder List

class Solution(object):
    def reorderList(self, head):
        
        # converting Linked List to list but keep whole single node in list not the value  
        a=[]
        c=head
        while c:
            a.append(c)
            c=c.next
        
        # applying the reording [1,2,3,4] -> [1,4,2,3] by two pointer approach
        i=0 # left starting 
        j=len(a)-1 # right at end 
        while i < j: 
            
            a[i].next = a[j]  # make [1] means adding value to next node 
            i+=1 
            
            if i==j: # if it matches break rotate complete 
                break
            
            a[j].next=a[i]  # make [1,4] means adding value to next node 
            j-=1
            
        a[i].next=None # if list over make last node None
        
        return head
```
