```python
# 724. Find Pivot Index Optimal Code

def pivotIndex(nums):
        g = sum(nums)
        t = 0
        for i, n in enumerate(nums):
            if t == g - t - n:
                return i
            t += n
        return -1
pivotIndex([1, 7, 3, 6, 5, 6])
```

```python
# 560. Subarray Sum Equals K Brute Force Code

def subarraySum(nums, k):
        c=0
        for l in range(len(nums)):
            s=0
            for r in range(l,len(nums)):
                s+=nums[r]
                if s == k:
                    c+=1
        return c
subarraySum([1,3,2,4,1,4,2,1,5],2)
```

```python
# 560. Subarray Sum Equals K Prefix Sum logic Optimal Using Hashmap

def subarraySum(nums, k):
        s={}
        s[0]=1
        c=0
        sum1=0
        
        for i in range(len(nums)):
            sum1 +=nums[i]
            c += s.get(sum1-k,0)
            s[sum1] =  s.get(sum1,0) +1
        return c
subarraySum([1,3,2,4,1,4,2,1,5],2)
```

```python
# 1314. Matrix Block Sum

def matrixBlockSum(mat, k):

        m, n = len(mat), len(mat[0])

        pre = [[0] * (n + 1) for _ in range(m + 1)]

        for i in range(m):
            for j in range(n):
                pre[i+1][j+1] = mat[i][j] + pre[i][j+1] + pre[i+1][j] - pre[i][j]

        ans = [[0] * n for _ in range(m)]

        for i in range(m):
            for j in range(n):

                r1 = max(0, i-k)
                c1 = max(0, j-k)

                r2 = min(m-1, i+k)
                c2 = min(n-1, j+k)

                ans[i][j] = (pre[r2+1][c2+1]
                           - pre[r1][c2+1]
                           - pre[r2+1][c1]
                           + pre[r1][c1])

        return ans
    
m=[[1,2,3],[4,5,6],[7,8,9]]
matrixBlockSum(m,1)
```

```python
# 238. Product of Array Except Self

def productExceptSelf(nums):
        l=len(nums)
        a=[0]*l
        a[0]=1
        for i in range(1,l):
            a[i] =nums[i-1]*a[i-1]
        r=1
        for i in range(l-1,-1,-1):
            a[i] *=r
            r*=nums[i]
        return a
n=[1,2,3,4]
print(productExceptSelf(n))
```

```python
# 523. Continuous Subarray Sum

def checkSubarraySum(nums, k):
        p=0
        m={0:-1}
        for i in range(len(nums)):
            p=(p+nums[i]) %k
            if p in m:
                if i-m[p] >1:
                    return True
            else:
                m[p]=i
        return False
checkSubarraySum([23,2,4,6,7],6)    
```

```python
# 974. Subarray Sums Divisible by K

def subarraysDivByK(nums, k):
        p = 0
        count = 0
        c = {0: 1}
        for i in nums:
            p += i
            r = p % k
            count += c.get(r, 0)
            c[r] = c.get(r, 0) + 1
        return count
subarraysDivByK([4,5,0,-2,-3,1],5)
```
