```python
# 232. Implement Queue using Stacks

class MyQueue(object):

    def __init__(self):
        self.s=[]
        self.s2=[]
        

    def push(self, x):
        while len(self.s) > 0:
            self.s2.append(self.s.pop())
        self.s.append(x)
        while len(self.s2) > 0:
            self.s.append(self.s2.pop())
            

    def pop(self):
        x=self.s[-1]
        self.s.pop()
        return x
        

    def peek(self):
        return self.s[-1]
        

    def empty(self):
        return len(self.s) ==0
        
```

```python
# 225. Implement Stack using Queues

from collections import deque
class MyStack(object):

    def __init__(self):
        self.q = deque()
        self.q2 = deque()
        

    def push(self, x):
        self.q2.append(x)
        while self.q:
            self.q2.append(self.q.popleft())
        self.q, self.q2 = self.q2, self.q
        
        

    def pop(self):
        return self.q.popleft()
        

    def top(self):
        return self.q[0]
        

    def empty(self):
        return len(self.q) == 0
```
