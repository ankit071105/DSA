```python
# stack implement using linkedlist

class Node:
    def __init__(self, d):
        self.d = d
        self.next = None


class Stack:
    def __init__(self):
        self.top = None
        self.count = 0

    def push(self, x):
        self.count += 1

        if self.top is None:
            self.top = Node(x)
            return

        n = Node(x)
        n.next = self.top
        self.top = n

    def pop(self):
        if self.top is None:
            return -1

        self.count -= 1
        x = self.top.d
        self.top = self.top.next
        return x

    def getTop(self):
        if self.top is None:
            return -1
        return self.top.d

    def size(self):
        return self.count
    
    
s = Stack()

s.push(2)
s.push(3)
s.push(4)
s.push(5)

print(s.pop())
print(s.pop())

print(s.size())
print(s.getTop())
    
    
        
        
        
        
```
