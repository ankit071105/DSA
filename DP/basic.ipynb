# DP

 1 Recursion Top-Down TC -> O(2^n) , SC -> O(N) (Stack space)

 2 Memoization Top-Down TC -> O(N) , SC -> O(N)+O(N) DP(Stack + Array)

 3 Tabulation Bottom - Up TC -> O(N) , SC -> O(N) DP(Array)

 4 Tabulation Space - Optimization TC -> O(N) , SC -> O(1)

#### nth Fibnoacci no

```python
# Recursive code for Fibnoacci No
def fib(n):
    if n==0:
        return 0
    if n==1:
        return 1
    return fib(n-1) + fib(n-2)
print(fib(5))
```

```python
# memoization code for Fibnoacci No
def fib(n):
    d=[-1]*(n+1)
    def solve(n):
        if n==0:
            return 0
        if n==1:
            return 1
        if d[n]!= -1:
            return d[n]
        d[n]=solve(n-1) + solve(n-2)
        return d[n]
    return solve(n)

print(fib(5))
```

```python
# Tabulation code for Fibnoacci No
def fib(n):
    if n==0:
        return 0
    if n==1:
        return 1
    d=[-1]*(n+1)
    d[0]=0
    d[1]=1
    for i in range(2,n+1):
        d[i]=d[i-1]+d[i-2]
    return d[n]
print(fib(5))
```

```python
# Tabulation (SPACE OPTIMIZATION) code for Fibnoacci No
def fib(n):
    if n==0:
        return 0
    if n==1:
        return 1
    p=0
    p1=1
    for i in range(2,n+1):
        c=p1+p
        p,p1=p1,c
    return c
print(fib(5))
```

#### Frog Jump

```python
# Recursive
def minCost(height):
        # code here
    def func(i):
        if i==0:
            return 0
        j=func(i-1)+ abs(height[i]-height[i-1])
        if i >1:
            j2= func(i-2)+abs(height[i]-height[i-2])
        else:
            j2=float('inf')
        return min(j,j2)
    return func(len(height) - 1)
heights = [20, 30, 40, 20]
print(minCost(heights))
```

```python
# memoization :-
def mincost(heights):
    n=len(heights)
    d=[-1]*n
    def func(i):
        if i==0:
            return 0
        if d[i] != -1:
            return d[i]
        j=func(i-1)+ abs(heights[i]-heights[i-1])
        if i >1:
            j2= func(i-2)+abs(heights[i]-heights[i-2])
        else:
            j2=float('inf')
        d[i]=min(j,j2)
        return d[i]
    return func(len(heights) - 1)
heights = [20, 30, 40, 20]
print(minCost(heights))
```

```python
# Tabulation :-
def mincost(heights):
    n=len(heights)
    d = [0] * n
    d[0]=0
    for i in range(1, n):
        one = d[i - 1] + abs(heights[i] - heights[i - 1])
        two = float('inf')
        if i > 1:
            two = d[i - 2] + abs(heights[i] - heights[i - 2])

        d[i] = min(one, two)

    return d[n - 1]
heights = [20, 30, 40, 20]
print(minCost(heights))
```

```python
# Tabulation (SPACE OPTIMIZATION) Code
def minCost(heights):
        n = len(heights)
        # base conditions
        if n == 1:
            return 0
        
        b = 0
        a = abs(heights[1] - heights[0])
        for i in range(2, n):
            one = a + abs(heights[i] - heights[i - 1])
            two = b + abs(heights[i] - heights[i - 2])
            # updating the answer
            curr = min(one, two)
            # swap 
            b, a = a, curr
        return a
heights = [20, 30, 40, 20]
print(minCost(heights))
```

### 198. House Robber

```python
# Recursive
def rob(nums):
        def func(i):
            if i==0:
                return nums[i]
            if i<0:
                return 0
            p=nums[i]+func(i-2)
            n=0+func(i-1)
            return max(p,n)
        return func(len(nums)-1)
nums = [1,2,3,1]
print(rob(nums))
        
```

```python
# memoization
def rob(nums):
        d=[-1]*(len(nums)+1)
        def func(i):
            if i==0:
                return nums[i]
            if i<0:
                return 0
            if d[i] != -1:
                return d[i]
            p=nums[i]+func(i-2)
            n=0+func(i-1)
            d[i]=max(p,n)
            return d[i]
        return func(len(nums)-1)
nums = [1,2,3,1]
print(rob(nums))
        
```

```python
# tabulation 
def rob(self, nums):
        n = len(nums)
        if n == 1:
            return nums[0]
        d = [0] * n
        d[0] = nums[0]
        for i in range(1, n):
            p = nums[i]
            if i > 1:
                p += d[i - 2]
            nu = d[i - 1]
            d[i] = max(p, nu)
        return d[n-1]
```

```python
# tabulation SPACE OPTIMIZATION
def rob(self, nums):
        n = len(nums)
        if n == 1:
            return nums[0]
        p=0
        p1=nums[0]
        for i in range(1, n):
            t = nums[i]+p
            nu=p1
            c=max(t,nu)
            p,p1=p1,c
            
        return p1
```
