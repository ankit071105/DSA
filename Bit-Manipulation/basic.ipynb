```python
# decimal to binary
def convert(n):
    r=''
    while n > 0:
        if n % 2 == 1:
            r+='1'
        else:
            r+='0'
        n//=2
    r=r[::-1]
    return r
print(convert(113))
```

```python
# binary to decimal
def convert(s):
    d=0
    p=0
    i=len(s)-1
    while i >=0:
        n=int(s[i]) * (2**p)
        d +=n
        i-=1
        p+=1
    return d
print(convert('1110001'))
        
```

```python
# Swap two no without third variable
def swap(a,b):
    a=a^b
    b=a^b
    a=a^b
    print(a,b)
swap(12,35)
    
```

```python
# check if the ith bit is set or not 
n = 13
i = 1
if n & (1<<i) ==0:
    print("False")
else:
    print("True")
```

```python
# set the ith  bit
n = 13
i = 2
x=n | (1<<i)
print(x)
```

```python
# clear the ith bit 
n = 13
i = 2
x= n & ~(1<<i)
print(x)
```
