```python
# 2. Add Two Numbers
class Solution(object):
    def addTwoNumbers(self, l1, l2):
        # converting linkedlist to list 
        c=l1
        a=[]
        while c:
            a.append(c.val)
            c=c.next
        b=[]
        c=l2
        while c:
            b.append(c.val)
            c=c.next
        d = []
        
        # add sum logic make it digit based addition 
        carry = 0
        for i in range(max(len(a), len(b))): # max size of list 
            x = a[i] if i < len(a) else 0
            y = b[i] if i < len(b) else 0
            t = x + y + carry # adding sum of digit and carry too
            d.append(t % 10) # add digit 
            carry = t // 10 #divide to get more digit 

        if carry: # if any carry is there then add to exisiting list 
            d.append(carry)
        
        # convert back to new linkedlist
        e = ListNode(0)
        c = e
        for i in d:
            c.next = ListNode(i)
            c = c.next
        return e.next
```

```python
# 1019. Next Greater Node In Linked List

# Input: head = [2,7,4,3,5]
# Output: [7,0,5,5,0]


class Solution(object):
    def nextLargerNodes(self, head):
        a=[]
        while head:
            a.append(head.val)
            head=head.next
        ans=[0]*len(a)
        s=[]
        for i in range(len(a)):
            while s and a[i] > a[s[-1]]:
                ans[s.pop()] = a[i]
            s.append(i)
        return ans
```

```python
# LeetCode 2487 - Remove Nodes From Linked List. Brute Force O(N^2) 
class Solution(object):
    def removeNodes(self, head):
        dummy=ListNode(0)
        prev=dummy
        dummy.next=head
        c=head
        while c:
            t=c.next
            d=False
            while t:
                if t.val > c.val:
                    d=True
                    break
                t=t.next
            if d== True:
                prev.next=c.next
            else:
                prev=c
            c=c.next
        return dummy.next
```

```python
# LeetCode 2487 - Remove Nodes From Linked List. Optimal SOlution O(n) O(1)  Stack
class Solution(object):
    def removeNodes(self, head):
        s=[]
        while head:
            while s and s[-1].val < head.val:
                s.pop()
            s.append(head)
            head=head.next
        for i in range(len(s)-1):
            s[i].next=s[i+1]
        s[-1].next=None
        return s[0]
```
